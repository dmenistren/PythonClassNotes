```python
# Indexing 

# postive - 0 to- + infinity
# negative - -1 to - infinity

# postive index
sample_text = "hello, I a"
              #0123456789  

sample_text[9]


# Negative index 
sample_text = "hello, I a"
             #10987654321  

sample_text[-7]

```

```python
# List indexing 

# postive and negative index

#       -3 -2 -1
value = [1, 2, 3]
#        0  1  2

value[-2]
```
